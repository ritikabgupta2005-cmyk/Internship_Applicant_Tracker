```python
#class
class InternshipApplication:

    #Constructor
    def __init__(self, app_id, company, role, location, stipend, status):
        self.app_id = app_id
        self.company = company
        self.role = role
        self.location = location
        self.stipend = stipend
        self.status = status

    #to display application details
    def display(self):
        print("\nApplication ID:", self.app_id)
        print("Company:", self.company)
        print("Role:", self.role)
        print("Location:", self.location)
        print("Stipend:", self.stipend)
        print("Status:", self.status)
```

```python
# objects
applications = [InternshipApplication(1,"TCS","Data Analyst Intern","Pune",15000,"Applied"),
    InternshipApplication(2,"Infosys","Python Intern","Bangalore",12000,"Interview"),
    InternshipApplication(3,"Accenture","Business Analyst Intern","Mumbai",18000,"Selected"),
    InternshipApplication(4,"Wipro","Data Science Intern","Hyderabad",16000,"Rejected"),
    InternshipApplication(5,"Deloitte","Finance Intern","Mumbai",14000,"Applied")]
```

```python
def display_applications():
    if len(applications) == 0:
        print("No applications available.")
        return
    for app in applications:
        app.display()
```

```python
def add_application():
    company = input("Enter company name: ")
    role = input("Enter internship role: ")
    location = input("Enter location: ")
    stipend = int(input("Enter stipend: "))
    status = input("Enter application status a)Applied b)Interview c)Selected d)Rejected: ")
    new_id = len(applications) + 1
    new_application = InternshipApplication(new_id,company,role,location,stipend,status)
    applications.append(new_application)
    print("Application added successfully!")
```

```python
def search_application():
    company = input("Enter company name to search: ")
    found = False
    for app in applications:
        if app.company.lower() == company.lower():
            print("\nApplication Found")
            print("ID:", app.app_id)
            print("Company:", app.company)
            print("Role:", app.role)
            print("Location:", app.location)
            print("Stipend:", app.stipend)
            print("Status:", app.status)
            found = True
    if found == False:
        print("Application not found.")
```

```python
def filter_by_status():
    status = input("Enter status: ")
    found = False
    for app in applications:
        if app.status.lower() == status.lower():
            print("\nCompany:", app.company)
            print("Role:", app.role)
            print("Location:", app.location)
            print("Stipend:", app.stipend)
            print("Status:", app.status)
            found = True
    if found == False:
        print("No applications found with this status.")
```

```python
def update_status():
    app_id = int(input("Enter application ID: "))
    new_status = input("Enter new status: ")
    for app in applications:
        if app.app_id == app_id:
            app.status = new_status
            print("Application status updated successfully!")
            return
    print("Application not found.")
```

```python
def highest_stipend():
    if len(applications) == 0:
        print("No applications available.")
        return
    highest = applications[0]
    for app in applications:
        if app.stipend > highest.stipend:
            highest = app
    print("\nHighest Stipend Application")
    print("Company:", highest.company)
    print("Role:", highest.role)
    print("Stipend:", highest.stipend)
```

```python
def average_stipend():
    if len(applications) == 0:
        print("No applications available.")
        return
    total = 0
    for app in applications:
        total = total + app.stipend
    average = total / len(applications)
    print("Average Stipend =", average)
```

```python
def count_by_status():
    applied = 0
    interview = 0
    selected = 0
    rejected = 0
    for app in applications:
        if app.status.lower() == "applied":
            applied += 1
        elif app.status.lower() == "interview":
            interview += 1
        elif app.status.lower() == "selected":
            selected += 1
        elif app.status.lower() == "rejected":
            rejected += 1
    print("\nApplication Summary")
    print("Applied:", applied)
    print("Interview:", interview)
    print("Selected:", selected)
    print("Rejected:", rejected)
```

```python
def delete_application():
    app_id = int(input("Enter application ID to delete: "))
    for app in applications:
        if app.app_id == app_id:
            applications.remove(app)
            print("Application deleted successfully!")
            return
    print("Application not found.")
```

```python
while True:
    print("\nINTERNSHIP APPLICANT TRACKER")
    print("1. Display All Applications")
    print("2. Add New Application")
    print("3. Search Application")
    print("4. Filter Applications by Status")
    print("5. Update Application Status")
    print("6. Find Highest Stipend")
    print("7. Calculate Average Stipend")
    print("8. Count Applications by Status")
    print("9. Delete Application")
    print("10. Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        display_applications()

    elif choice == "2":
        add_application()

    elif choice == "3":
        search_application()

    elif choice == "4":
        filter_by_status()

    elif choice == "5":
        update_status()

    elif choice == "6":
        highest_stipend()

    elif choice == "7":
        average_stipend()

    elif choice == "8":
        count_by_status()

    elif choice == "9":
        delete_application()

    elif choice == "10":
        print("Thank you for using Internship Applicant Tracker!")
        break

    else:
        print("Invalid choice. Please try again.")
```
